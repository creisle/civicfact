```python
from article_utils.util import read_jsonl
from figutils import add_bar_labels
import glob
import seaborn as sns
import pandas as pd
from bs4 import BeautifulSoup



def get_evidence_from_html(html):
    soup = BeautifulSoup(html)
    ev = soup.select_one('.evidence > p')
    return ev.get_text()

other_df = read_jsonl(*[f for f in glob.glob('data/*.jsonl') if f != 'data/all.jsonl'], add_filename=True).rename(columns={'id': 'doccano.id', 'text': 'doccano.text'})
print(other_df.columns)
meta_df = read_jsonl('data/all.jsonl')
print('other_df', other_df.shape[0])

df = other_df.merge(meta_df)
assert df.shape[0] == other_df.shape[0]
print('df', df.shape[0])
print(df.columns.tolist())
df['label'] = df.label.apply(lambda labels: labels[0] if labels else '')
df['predicted_label_name'] = df.label.str.extract(r'^(\w+)')
df['predicted_label_name'] = df.predicted_label_name.str.upper()
df['curator'] = df.filename.str.extract('data/(\w+)\.jsonl')

df['civic.evidenceDirection'] = df['civic.evidenceDirection'].fillna('SUPPORTS')
df.loc[df['civic.evidenceDirection'] == '', 'civic.evidenceDirection'] = 'SUPPORTS'
df.loc[df['civic.evidenceDirection'] == 'DOES_NOT_SUPPORT', 'civic.evidenceDirection'] = 'REFUTES'

df['freq'] = 1

agg_df = df.groupby(['document.pmid', 'display_id', 'civic.evidenceDirection', 'gold_label_name',  'correct.strict']).agg({'predicted_label_name': list, 'freq': 'sum'}).reset_index().rename(columns={'predicted_label_name': 'labels'})
agg_df['majority.predicted_label_name'] = agg_df.labels.apply(lambda labels: max(set(labels), key=labels.count))
agg_df['majority.predicted_label_support'] = agg_df.apply(lambda row: sum([1 for l in row.labels if l == row['majority.predicted_label_name']]) / len(row.labels), axis=1)
agg_df['majority.num_labels'] = agg_df.labels.str.len()
agg_df.loc[agg_df['majority.predicted_label_support'] <= 0.5, 'majority.predicted_label_name'] = 'NO MAJORITY'
agg_df = agg_df.rename(columns={'predicted_label_name': 'majority.predicted_labels'}).drop(columns=['freq'])

df = df.merge(agg_df)

print(df.model_name.unique())

df['example_key'] = df['claim.flat'] + '' + df['evidence.flat'] + df['document.pmid'].astype(str)
print(f'There are {df.example_key.nunique()} unique examples in the review dataset')


def detect_merging_error(evidence):
    for ev in evidence:
        if (ev['meta']['end'] - ev['meta']['start']) != len(ev['content']):
            return True
    return False


# what examples were affected by the chunk merging bug?
df['old_evidence.flat'] = df['doccano.text'].apply(get_evidence_from_html)
df['merging_error'] = df['evidence'].apply(detect_merging_error)

fig = sns.countplot(agg_df, x='majority.predicted_label_support')
add_bar_labels(fig)
df
```

```python
# # How much of the expected evidence charaters were in the output?
from nli.util import concat_text

def compute_overlap(range1, range2):
    overlap = min(range1[1], range2[1]) - max(range1[0], range2[0])
    if overlap < 0:
        return 0
    return overlap

def expected_evidence_coverage(row):
    coverage = 0
    for ev_group in row['all_expected_evidence']:
        for expected_chunk in ev_group:
            for obs_chunk in row['evidence']:
                overlap = compute_overlap((expected_chunk['meta']['start'], expected_chunk['meta']['end']), (obs_chunk['meta']['start'], obs_chunk['meta']['end']))
                coverage += overlap

    return coverage

def expected_evidence_length(row):
    total_expected = 0
    for ev in row['all_expected_evidence']:
        for expected_chunk in ev:
            total_expected += expected_chunk['meta']['end'] - expected_chunk['meta']['start']

    return total_expected

build_df = read_jsonl('../../../builds/civicfact-2024.12.22/data.jsonl.gz')
build_df = build_df.groupby(['claim.flat', 'document.pmid']).agg({'evidence': list}).reset_index().rename(columns={'evidence': 'all_expected_evidence'})
pre_size = df.shape[0]
if 'all_expected_evidence' in df.columns:
    df = df.drop(columns=['all_expected_evidence', 'ret_coverage'])
df = df.merge(build_df, on=['claim.flat', 'document.pmid'])
assert pre_size == df.shape[0]
df['expected_evidence.cover'] = df.apply(expected_evidence_coverage, axis=1).round(1)
df['expected_evidence.length'] = df.apply(expected_evidence_length, axis=1).round(1)
df['expected_evidence.coverage'] = df['expected_evidence.cover'] / df['expected_evidence.length']
df['expected_evidence.flat'] = df['expected_evidence'].apply(concat_text)
df['expected_evidence.ranges'] = df['expected_evidence'].apply(lambda ev: ' '.join(f'{e["meta"]["start"]}-{e["meta"]["end"]}' for e in ev))
df['evidence.ranges'] = df['evidence'].apply(lambda ev: ' '.join(f'{e["meta"]["start"]}-{e["meta"]["end"]}' for e in ev))

df = df.merge(df.groupby(['display_id']).agg({'expected_evidence.coverage': 'max'}).reset_index().rename(columns={'expected_evidence.coverage': 'ret_coverage'}))


sns.displot(df.drop_duplicates(['tracking_id']), x='ret_coverage', kind='hist')
```

## Inter-Annotator Agreement Metrics

For all IA caluclations we will exlude the curator who dropped out, malachig

```python
from sklearn.metrics import cohen_kappa_score
import itertools
import numpy as np
from figutils import plot_heatmap

curators = [c for c in df.curator.unique().tolist() if c != 'malachig']
scores = []
np.seterr(all='raise')
MIN_SHARED_EXAMPLES = 3


for (curator1, curator2) in itertools.product(curators, repeat=2):
    cur1_df = df[df.curator == curator1].rename(columns={'predicted_label_name': 'curator1'})[['curator1', 'display_id']]
    cur2_df = df[df.curator == curator2].rename(columns={'predicted_label_name': 'curator2'})[['curator2', 'display_id']]

    cur_df = cur1_df.merge(cur2_df)
    if cur_df.shape[0] < MIN_SHARED_EXAMPLES:

        print('too few examples', cur_df.shape[0], 'to calculate agreement between', curator1, 'and', curator2)
        continue
    score = cohen_kappa_score(cur_df['curator1'].tolist(), cur_df['curator2'].tolist())
    scores.append({'curator1': curator1, 'curator2': curator2, 'score': score, 'shared_examples_count': cur_df.shape[0]})

scores_df = pd.DataFrame.from_records(scores)
print('average score:', scores_df[scores_df.curator1 != scores_df.curator2].score.mean())
print('average count:', scores_df[scores_df.curator1 != scores_df.curator2].shared_examples_count.mean())
scores_df['w_score'] = scores_df.shared_examples_count * scores_df.score
weight_avg = scores_df.w_score.sum() / scores_df.shared_examples_count.sum()
print('weighted avg:', weight_avg)
fig = sns.catplot(scores_df[scores_df.curator1 != scores_df.curator2], y='curator1', x='score', kind='box', weights='shared_examples_count' )
fig.set(xlabel='cohen kappa score')
plot_heatmap(scores_df, rows_col='curator1', cols_col='curator2', data_col='score', col_cluster=True, row_cluster=True)
```

## Correlation

None of the above metrics take into account that NEI vs SUPPORTS is less of a disagreement than REFUTES vs SUPPORTS. If we map the labels to a number scale we can use correlation to take this into account (see below)

- REFUTES: -1
- NEI: 0
- SUPPORTS: +1

```python
from scipy.stats import spearmanr

label_values = {'NEI': 0, 'SUPPORTS': 1, 'REFUTES': -1}

df['predicted_label_value'] = df['predicted_label_name'].apply(lambda name: label_values[name])


curators = [c for c in df.curator.unique().tolist() if c != 'malachig']
scores = []

for (curator1, curator2) in itertools.product(curators, repeat=2):
    cur1_df = df[df.curator == curator1].rename(columns={'predicted_label_value': 'curator1'})[['curator1', 'display_id']]
    cur2_df = df[df.curator == curator2].rename(columns={'predicted_label_value': 'curator2'})[['curator2', 'display_id']]

    cur_df = cur1_df.merge(cur2_df)
    if cur_df.shape[0] < MIN_SHARED_EXAMPLES:

        print('too few examples', cur_df.shape[0], 'to calculate agreement between', curator1, 'and', curator2)
        continue
    if curator1 == curator2:
        score = 1
    else:
        stat = spearmanr(cur_df['curator1'].tolist(), cur_df['curator2'].tolist())
        score = stat.statistic

    scores.append({'curator1': curator1, 'curator2': curator2, 'score': score, 'shared_examples_count': cur_df.shape[0]})

scores_df = pd.DataFrame.from_records(scores)
print(scores_df[scores_df.curator1 != scores_df.curator2]['score'].mean())
print(scores_df['score'].unique())

plot_heatmap(scores_df, rows_col='curator1', cols_col='curator2', data_col='score')
```
