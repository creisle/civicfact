```python
# now match up the spreadsheet to create the labels
def extract_new_labels(row):
    labels = set()
    if row['problem_status'] == 'not FT':
        labels.add('not in FT: no FT')
    if row['problem_status'] == 'ok':
        labels.add('civic change: none')
    status_map = {
        'already rejected': 'civic change: already rejected',
        'ok-ish': 'ok-ish',
        'already pending revisions': 'civic change: already pending revisions',
        'ok': 'civic change: none',
    }
    if row['status'] in status_map:
        labels.add(status_map[row['status']])

    action_map = {
        'disease specificity error': 'specificity error: disease',
        'drug specificity': 'specificity error: drug',
        'model update': 'civic change: civic model update',
        'reject': 'civic change: should be flagged or rejected',
        'therapy interaction type': 'specificity error: drug',
        'variant specificity error': 'specificity error: variant',
        'wrong disease term': 'wrong: disease',
        'wrong phenotypes': 'wrong: phenotype',
        'wrong significance': 'wrong: significance',
        'wrong variant type': 'wrong: variant',
        'already rejected': 'civic change: already rejected',
        'complex MP': 'civic change: should be combination variant',
        'data model change update': 'civic change: civic model update',
        'wrong drug included': 'wrong: drug',
        'wrong evidence direction': 'wrong: evidence direction',
        'wrong evidence type': 'wrong: evidence type',
        'wrong variant': 'wrong: variant',
        'rejected': 'civic change: should be flagged or rejected',
    }
    if not pd.isnull(row['action']):
        for action in row['action'].split(';'):
            action = action.strip()
            if action not in action_map:
                print(f'warning: no mapping for this action value ({action})')
            else:
                labels.add(action_map[action])

    return labels

df['label'] = df.apply(extract_new_labels, axis=1)



doc_df = doc_df.drop(columns=['label']).merge(df[['civic.id', 'label', 'notes', 'new gold label']], how='left')
# dump_jsonl(doc_df, 'data/doccano/2025.11.17_civicreview_v2025.03.25/inputs/doccano_input.v1.with_labels.jsonl.gz')

for _, row in df[(~df.notes.isna()) | (~df['new gold label'].isna())].reset_index().iterrows():
    print()
    for field in ['civic.id', 'document.pmid', 'notes', 'new gold label']:
        print(field + ': ', row[field])
```
